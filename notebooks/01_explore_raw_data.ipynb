```python
all_df = pl.read_parquet("../data/raw/daily/*.parquet")   # * 表示读取所有文件

check = (
    all_df.sort("code", "date")
    .with_columns(prev_close=pl.col("close").shift(1).over("code"))   # 每只股票各自取上一行的收盘价
    .with_columns(ratio=pl.col("preclose") / pl.col("prev_close"))
)

check.filter(pl.col("ratio") < 0.95).select(
    "date", "code", "prev_close", "preclose", "close", "pct_chg", "ratio"
)
```

```python
check.filter(pl.col("tradestatus") == 0).select(
    "date", "code", "prev_close", "open", "high", "low", "close", "volume", "amount", "turn"
).head(20)
```
